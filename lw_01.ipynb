```python
# Лабораторная работа 1
# Задание 12: Симуляция очереди печати принтера (добавление задач и "печать").
#
# Выполнил: Князева А. П.
# Группа: ЦИБ-251

printer_queue = []

print("Симуляция очереди печати принтера")

while True:
    print("\nОчередь печати:", printer_queue)
    print("1 - Добавить задачу")
    print("2 - Печать")
    print("3 - Выход")

    choice = input("Выберите действие (1-3): ")

    if choice == "1":
        task = input("Введите имя документа: ")
        printer_queue.append(task)
        print(f"Добавлено: {task}")

    elif choice == "2":
        if printer_queue:
            task = printer_queue.pop(0)
            print(f"Печатается: {task}")
        else:
            print("Нет задач в очереди")

    elif choice == "3":
        print("Программа завершена. До свидания!")
        break
```
