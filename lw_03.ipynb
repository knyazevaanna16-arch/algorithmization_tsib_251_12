```python
# Задание task_03_12.

# Цены конкурентов (Бизнес-задача)

''' Дан список цен конкурентов prices (float). Маркетологам нужно отсортировать их
по возрастанию, чтобы найти минимальные предложения. Используйте
сортировку выбором в функции sort_competitor_prices(prices). '''

# Выполнил: Князева А. П.
# Группа: ЦИБ-251

def sort_competitor_prices(prices):      # O(1) - объявление функции
    n = len(prices)                      # O(1) - получение длины списка
    for i in range(n - 1):               # O(n) - внешний цикл выполняется n-1 раз
        min_index = i                    # O(1) - присваивание внутри цикла
        for j in range(i + 1, n):        # O(n) - внутренний цикл (в среднем ~n/2 итераций)
            if prices[j] < prices[min_index]:  # O(1) - сравнение двух элементов
                min_index = j            # O(1) - присваивание индекса
        if min_index != i:               # O(1) - проверка условия после внутреннего цикла
            prices[i], prices[min_index] = prices[min_index], prices[i]  # O(1) - обмен элементами

# Общая временная сложность данного кода составляет O(n^2), где n — количество элементов в списке prices.

if __name__ == "__main__":
    a = [99.9, 45.0, 50.0]
    sort_competitor_prices(a)
    print(a)

    a = [500.0, 100.0]
    sort_competitor_prices(a)
    print(a)
```
