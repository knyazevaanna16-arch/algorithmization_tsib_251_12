```python
# Задание task_04_12.

# Простая верификация. Функция verify_salted(stored_hash, password, salt)
# проверяет, совпадает ли хеш введенного пароля с сохраненным.

# Выполнил: Князева А. П.
# Группа: ЦИБ-251

import hashlib

def verify_salted(stored_hash, password, salt):
    # Склеиваем пароль и соль
    combined = password + salt
    # Считаем SHA-256
    hash_object = hashlib.sha256(combined.encode())
    calculated_hash = hash_object.hexdigest()
    # Сравниваем
    return calculated_hash == stored_hash

if __name__ == "__main__":

    correct_password = "pluto89206"
    salt = "mareon"

    # Получаем хеш, который хранится в базе
    stored_hash = hashlib.sha256((correct_password + salt).encode()).hexdigest()
    print(f"Хеш в базе: {stored_hash}")
    print()

    # Бесконечный цикл для многократного ввода
    while True:
        user_input = input("Введите пароль (или 'exit' для выхода): ")

        # Условие выхода из цикла
        if user_input.lower() == 'exit':
            print("Программа завершена.")
            break

        # Проверка пароля
        result = verify_salted(stored_hash, user_input, salt)

        if result:
            print("Результат: True (Пароль верный! Доступ разрешен.)")
            print()
        else:
            print("Результат: False (Пароль неверный. Доступ запрещен.)")
            print()
```
